```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Correlation:", round(df["search_volume"].corr(df["impressions_90d"]), 3))
print("Average CTR:", round(df["ctr"].mean(), 3))
print("Average Position:", round(df["avg_position"].mean(), 2))
```

## Quick Look at the Data

The starter dataset contains **30,000 web pages** and **44 features**, providing enough information to analyze search performance.

The correlation between **search volume** and **impressions** is only **0.001**, showing that search volume alone is not a strong indicator of page visibility.

The average **CTR is 0.511** and the average **search position is 16.34**, indicating that multiple ranking-related signals should be analyzed together rather than relying on a single metric.
